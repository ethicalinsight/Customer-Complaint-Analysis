#Data Understanding by Displaying Its Rows, Columns, Statistical and Concise Descriptions, Column Names, and Data Types
##To explore and understand your dataset thoroughly in Python using Pandas, you can use the following functions:

###Displaying Rows: Use df.head(n) to view the first n rows and df.tail(n) to view the last n rows of the DataFrame.

###Displaying Columns & Shape: Use df.shape to check the total number of rows and columns, and df.columns to retrieve all column names.

###Concise Descriptions: Use df.info() to get a concise summary of the DataFrame, including the non-null count and data types for every column.

###Statistical Descriptions: Use df.describe() to generate descriptive statistics (such as mean, standard deviation, min, max, and percentiles) for all numerical columns.
